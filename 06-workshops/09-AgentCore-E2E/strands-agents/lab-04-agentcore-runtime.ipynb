### AgentCore Runtime과 프로덕션 배포가 중요한 이유

현재 상태(실습 1~3): Agent는 중앙화된 도구를 사용하여 로컬에서 실행되지만 프로덕션 환경에서는 다음과 같은 문제가 있습니다.

- Agent가 단일 세션으로 로컬에서 실행됨
- 종합적인 모니터링 또는 디버깅 기능이 없음
- 여러 동시 사용자를 안정적으로 처리할 수 없음

이 실습을 마치면 다음 기능을 갖춘 프로덕션 지원 Agent 인프라를 구축하게 됩니다.

- 변동하는 수요를 처리하는 서버리스 자동 확장
- 추적, 지표, 로깅을 통한 종합적인 Observability
- 자동 오류 복구를 통한 엔터프라이즈급 안정성
- 적절한 액세스 제어를 적용한 안전한 배포
- AWS 콘솔과 API를 통한 간편한 관리 및 실제 프로덕션 워크로드 지원


### AgentCore Observability로 종합적인 Observability 추가

AgentCore Runtime은 [AgentCore Observability](https://docs.aws.amazon.com/bedrock-agentcore/latest/devguide/observability.html)와 원활하게 통합되어 프로덕션 환경의 Agent 동작을 완전히 파악할 수 있도록 합니다. AgentCore Observability는 Agent 상호 작용, 도구 사용, Memory 액세스 패턴에서 추적, 지표, 로그를 자동으로 캡처합니다. 이 실습에서는 AgentCore Runtime과 CloudWatch GenAI Observability를 통합하여 종합적인 모니터링 및 디버깅 기능을 제공하는 방법을 살펴봅니다.

요청 추적 시 AgentCore Observability는 도구 호출, Memory 검색, 모델 상호 작용을 포함한 전체 대화 흐름을 캡처합니다. 성능 모니터링 시에는 응답 시간, 성공률, 리소스 사용률을 추적하여 Agent 성능을 최적화할 수 있도록 지원합니다.

Observability 흐름에서 AgentCore Runtime은 Agent 코드를 자동으로 계측하고 텔레메트리 데이터를 CloudWatch로 전송합니다. 그런 다음 CloudWatch 대시보드와 GenAI Observability 기능을 사용하여 패턴을 분석하고 병목 현상을 식별하며 문제를 실시간으로 해결할 수 있습니다.

### 단계 2: AgentCore Runtime용 Agent 준비

#### Runtime 지원 Agent 생성

먼저 이전 로컬 Agent 구현에서 Python SDK를 통해 필요한 AgentCore Runtime 구성 요소를 정의합니다.

아래의 #### AGENTCORE RUNTIME - 줄 1 #### 주석을 살펴보면 관련 배포 코드가 추가된 위치를 확인할 수 있습니다. Runtime 지원 Agent를 준비하는 줄은 총 네 개입니다.

1. `from bedrock_agentcore.runtime import BedrockAgentCoreApp`으로 Runtime App 가져오기
2. `app = BedrockAgentCoreApp()`으로 App 초기화
3. 호출 함수에 `@app.entrypoint` 데코레이터 적용
4. `app.run()`으로 AgentCore Runtime이 실행을 제어하도록 설정

##### 주요 구현 세부 정보

Runtime 지원 Agent는 entrypoint 함수를 사용하여 payload에서 user prompt를 추출하고 
context.request_headers.get('Authorization', '')를 통해 요청 헤더에서 JWT 토큰을 추출합니다. 그런 다음 권한 부여 토큰을 
MCP 클라이언트 헤더 `headers={"Authorization": auth_header}`에 전달하여 AgentCore Gateway로 직접 전파합니다. 이 구현에는 인증 누락에 대한 오류 처리가 포함되어 있으며, 
이전 실습의 모든 Memory 및 도구 기능을 유지하면서 동기식 Agent 호출에서 일반 텍스트 응답을 반환합니다.

````python
%%writefile ./lab_helpers/lab4_runtime.py
import os
from bedrock_agentcore.runtime import (
    BedrockAgentCoreApp,
)  #### AGENTCORE RUNTIME - 줄 1 ####
from strands import Agent
from strands.tools.mcp import MCPClient
from mcp.client.streamable_http import streamablehttp_client
import requests
import boto3
from strands.models import BedrockModel
from lab_helpers.utils import get_ssm_parameter
from lab_helpers.lab1_strands_agent import (
    get_return_policy,
    get_product_info,
    get_technical_support,
    SYSTEM_PROMPT,
    MODEL_ID,
)

from lab_helpers.lab2_memory import (
    ACTOR_ID,
    SESSION_ID,
)
from bedrock_agentcore_starter_toolkit.operations.memory.manager import MemoryManager
from bedrock_agentcore.memory.integrations.strands.config import AgentCoreMemoryConfig, RetrievalConfig
from bedrock_agentcore.memory.integrations.strands.session_manager import AgentCoreMemorySessionManager

# boto3 클라이언트 초기화
sts_client = boto3.client('sts')

# AWS 계정 세부 정보 가져오기
REGION = boto3.session.Session().region_name

# 실습 1 가져오기: Bedrock 모델 생성
model = BedrockModel(model_id=MODEL_ID)

# 실습 2 가져오기: Memory
memory_id = os.environ.get("MEMORY_ID")
if not memory_id:
    raise Exception("Environment variable MEMORY_ID is required")

# AgentCore Runtime App 초기화
app = BedrockAgentCoreApp()  #### AGENTCORE RUNTIME - 줄 2 ####

@app.entrypoint  #### AGENTCORE RUNTIME - 줄 3 ####
async def invoke(payload, context=None):
    """AgentCore Runtime 엔트리포인트 함수입니다."""
    user_input = payload.get("prompt", "")
    session_id = context.session_id # context에서 session_id 가져오기
    actor_id = payload.get("actor_id", ACTOR_ID) 
    # 요청 헤더에 액세스 - None인 경우 처리
    request_headers = context.request_headers or {}

    # 클라이언트 JWT 토큰 가져오기
    auth_header = request_headers.get('Authorization', '')

    print(f"Authorization header: {auth_header}")
    # Gateway ID 가져오기
    existing_gateway_id = get_ssm_parameter("/app/customersupport/agentcore/gateway_id")
    
    # Bedrock AgentCore Control 클라이언트 초기화
    gateway_client = boto3.client(
        "bedrock-agentcore-control",
        region_name=REGION,
    )
    # 기존 Gateway 세부 정보 가져오기
    gateway_response = gateway_client.get_gateway(gatewayIdentifier=existing_gateway_id)

    # Gateway URL 가져오기
    gateway_url = gateway_response['gatewayUrl']

    # JWT 토큰이 있으면 context manager 내에서 MCP 클라이언트와 Agent 생성
    if gateway_url and auth_header:
        try:
                mcp_client = MCPClient(lambda: streamablehttp_client(
                    url=gateway_url,
                    headers={"Authorization": auth_header}  
                ))
                
                with mcp_client:
                    tools = (
                        [
                            get_product_info,
                            get_return_policy,
                            get_technical_support
                        ]
                        + mcp_client.list_tools_sync()
                    )

                    memory_config = AgentCoreMemoryConfig(
                        memory_id=memory_id,
                        session_id=str(session_id),
                        actor_id=actor_id,
                        retrieval_config={
                            "support/customer/{actorId}/semantic/": RetrievalConfig(top_k=3, relevance_score=0.2),
                            "support/customer/{actorId}/preferences/": RetrievalConfig(top_k=3, relevance_score=0.2)
                        }
                    )

                    # 모든 고객 지원 도구를 포함한 Agent 생성
                    agent = Agent(
                        model=model,
                        tools=tools,
                        system_prompt=SYSTEM_PROMPT,
                        session_manager=AgentCoreMemorySessionManager(memory_config, REGION),
                    )
                    # Agent 호출
                    response = agent(user_input)
                    return response.message["content"][0]["text"]
        except Exception as e:
                print(f"MCP client error: {str(e)}")
                return f"Error: {str(e)}"
    else:
        return "Error: Missing gateway URL or authorization header"

if __name__ == "__main__":
    app.run()  #### AGENTCORE RUNTIME - 줄 4 ####
````

#### AgentCore Runtime 구성 요약

아래 코드는 Starter Toolkit을 사용하여 AgentCore Runtime 배포를 구성합니다. Runtime용 실행 Role을 생성한 다음

Cognito를 사용하는 JWT 기반 인증을 설정합니다. 이 구성은 SSM 파라미터에서 가져온 허용 client ID 및 discovery URL을 지정하여 
프로덕션 Agent 배포에 안전한 액세스 제어를 설정합니다. 이 단계에서는 컨테이너 배포에 필요한 
Dockerfile과 .bedrock_agentcore.yaml 구성 파일을 자동으로 생성합니다.

**Runtime 헤더 구성**: 아래 코드는 배포된 AgentCore Runtime의 사용자 지정 헤더 허용 목록을 구성합니다. Agent ARN에서 Runtime ID를 추출하고, 
기존 설정을 유지하기 위해 현재 Runtime 구성을 가져온 다음, Authorization 헤더
(OAuth 토큰 전파에 필요)와 사용자 지정 헤더를 포함하는 요청 헤더 허용 목록으로 Runtime을 업데이트합니다. 이를 통해 JWT 토큰 및 기타 필수 헤더가 
클라이언트 요청에서 Agent Runtime 코드로 올바르게 전달됩니다.

**참고: 이 워크숍 범위에서는 Platform Mismatch 경고를 무시해도 됩니다.**

````python
from bedrock_agentcore_starter_toolkit import Runtime
from lab_helpers.utils import create_agentcore_runtime_execution_role

# Runtime Toolkit 초기화
boto_session = boto3.session.Session()
region = boto_session.region_name

execution_role_arn = create_agentcore_runtime_execution_role()

agentcore_runtime = Runtime()

# 배포 구성
response = agentcore_runtime.configure(
    entrypoint="lab_helpers/lab4_runtime.py",
    execution_role=execution_role_arn,
    auto_create_ecr=True,
    requirements_file="requirements.txt",
    region=region,
    agent_name="customer_support_agent",
    authorizer_configuration={
        "customJWTAuthorizer": {
            "allowedClients": [get_ssm_parameter("/app/customersupport/agentcore/client_id")],
            "discoveryUrl": get_ssm_parameter("/app/customersupport/agentcore/cognito_discovery_url"),
        }
    },
    # Authorization 및 사용자 지정 헤더 허용 목록 추가
    request_header_configuration={
        "requestHeaderAllowlist": [
            "Authorization",  # OAuth 전파에 필요
            "X-Amzn-Bedrock-AgentCore-Runtime-Custom-H1",  # 사용자 지정 헤더
        ]
    },
)

print("Configuration completed:", response)
````
