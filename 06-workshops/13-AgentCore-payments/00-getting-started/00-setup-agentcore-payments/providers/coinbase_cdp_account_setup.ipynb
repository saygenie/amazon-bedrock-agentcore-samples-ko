# Provider 설정: Coinbase Developer Platform(CDP)

## 개요

**Amazon Bedrock AgentCore payments**는 Coinbase Developer Platform(CDP)을 wallet
infrastructure로 사용합니다. CDP는 API를 통해 crypto wallet을 provision하고 관리하므로 에이전트가
private key를 직접 보유하지 않습니다.

이 가이드에서는 Coinbase 계정 생성, CDP 활성화, Tutorial 00에 필요한 세 가지
credentials 생성 과정을 안내합니다.

### 준비되는 항목

이 가이드를 마치면 `.env`에 추가할 세 가지 값이 준비됩니다.

| Variable | 설명 |
|:---------|:------------|
| `COINBASE_API_KEY_ID` | CDP API key 식별 |
| `COINBASE_API_KEY_SECRET` | CDP API 호출 인증 |
| `COINBASE_WALLET_SECRET` | CDP에서 관리하는 wallet 잠금 해제 |

**참고:** Wallet Secret은 생성 시 한 번만 표시됩니다. 즉시 안전한 위치에
저장하세요(AWS Secrets Manager, 1Password 등). 나중에는 다시 가져올 수 없습니다.

### 사전 요구 사항

* 유효한 email address
* SMS 인증용 phone number

### 예상 소요 시간

계정 생성에는 약 10분, CDP key 생성에는 약 5분이 걸립니다.

### 수행할 작업

| 단계 | 위치 | 작업 |
|:-----|:------|:-------|
| 1 | coinbase.com | 계정 생성(이미 있으면 건너뜀) |
| 2 | portal.cdp.coinbase.com | CDP project 생성 |
| 3a | CDP Portal → API Keys | API key 생성 → `API Key ID` + `API Key Secret` 복사 |
| 3b | CDP Portal → Wallets → ServerWallet | `Wallet Secret` 복사 |
| 4 | 이 Notebook | credentials 붙여넣기 → 셀이 `update_env_file()`을 통해 `.env`에 기록 |
| 5 | 이 Notebook(선택 사항) | credentials가 올바르게 load되는지 확인 |
| 6 | CDP Portal → Wallets → Embedded Wallet → Policies | ✋ Delegated Signing 활성화 |

1~6단계는 Tutorial 00보다 먼저 수행합니다. Wallet funding은 wallet이 생성된 후 Tutorial 00의 7b단계에서 수행합니다.

## 1단계 — Coinbase 계정 생성

1. **[coinbase.com](https://coinbase.com/)**으로 이동합니다.
2. **Get started**를 선택합니다.
3. email address를 입력합니다.
4. password를 생성합니다.
5. Coinbase에서 전송한 link를 통해 email을 인증합니다.
6. phone verification(SMS code)을 완료합니다.

**참고:** Coinbase 계정이 이미 있으면 2단계로 이동하세요.

## 2단계 — Coinbase Developer Platform 활성화

1. **[portal.cdp.coinbase.com](https://portal.cdp.coinbase.com/)**으로 이동합니다.
2. Coinbase 계정으로 로그인합니다.
3. **Project**를 생성하거나 기존 project를 선택합니다.

![CDP Portal — Project 페이지](../images/coinbase-setup/coinbase-setup-01.png)

이제 Coinbase Developer Platform에 접속했습니다. 필요한 주요 영역은
**API Keys**와 **Wallets**입니다.

## 3단계 — API Key 생성 및 Wallet Secret 가져오기

[CDP Portal](https://portal.cdp.coinbase.com/)에서 세 가지 credentials가 필요합니다. 처음 두 개는 API Keys에서, 세 번째는 Wallets 아래의 별도 섹션에서 가져옵니다.

### 3a. API Key 생성

1. CDP Portal에서 project로 이동합니다.
2. **Create API key** 또는 **New key**를 선택합니다.
3. 설명이 포함된 이름을 입력합니다(예: `agentcore-payments-tutorial`).
4. **Create** 또는 **Save**를 선택합니다.
5. 안내가 표시되면 key를 download합니다.

![CDP Portal — API Key 생성](../images/coinbase-setup/coinbase-setup-02.png)

다음 값을 얻을 수 있습니다.

| CDP Field | 대응하는 .env variable |
|:----------|:---------------------|
| API Key ID | `COINBASE_API_KEY_ID` |
| API Key Secret (Private key) | `COINBASE_API_KEY_SECRET` |

### 3b. Wallet Secret 가져오기

Wallet Secret은 cryptographic wallet operation에 사용되는 별도의 credential이며 API key와 다른 섹션에 있습니다.

1. CDP Portal에서 **Wallets** → **ServerWallet**로 이동합니다.
2. **Wallet Secret**(때로는 "wallet passphrase"라고 표시됨)을 복사합니다.

![CDP Portal — Wallet Secret](../images/coinbase-setup/coinbase-setup-03.png)

| CDP Field | 대응하는 .env variable |
|:----------|:---------------------|
| Wallet Secret | `COINBASE_WALLET_SECRET` |

**Wallet Secret은 한 번만 표시됩니다.** 즉시 안전한 위치에 저장하세요.

**팁:** 계속 진행하기 전에 세 값을 모두 AWS Secrets Manager 또는 password manager에 저장하세요.
절대로 git에 commit하지 마세요.

## 4단계 — `.env`에 Credentials 추가

아래에 Coinbase 값 세 개를 붙여넣고 셀을 실행합니다. 수동 편집 없이 `../.env`에 직접 기록됩니다.

Tutorial 00에서 사용할 provider를 알 수 있도록 `CREDENTIAL_PROVIDER_TYPE=CoinbaseCDP`를 설정합니다.

```python
import sys
import os

sys.path.append("../..")
from utils import update_env_file

# ✋ Coinbase CDP credentials를 여기에 붙여넣기
COINBASE_API_KEY_ID = "<paste your API Key ID here>"
COINBASE_API_KEY_SECRET = "<paste your API Key Secret here>"
COINBASE_WALLET_SECRET = "<paste your Wallet Secret here>"

# 기록 전 검증
for name, val in [
    ("COINBASE_API_KEY_ID", COINBASE_API_KEY_ID),
    ("COINBASE_API_KEY_SECRET", COINBASE_API_KEY_SECRET),
    ("COINBASE_WALLET_SECRET", COINBASE_WALLET_SECRET),
]:
    if not val or val.startswith("<"):
        raise ValueError(f"{name} is not set. Paste your value above and re-run.")

# .env에 기록
env_path = os.path.join("..", "..", ".env")
result = update_env_file(
    env_path,
    {
        "CREDENTIAL_PROVIDER_TYPE": "CoinbaseCDP",
        "COINBASE_API_KEY_ID": COINBASE_API_KEY_ID,
        "COINBASE_API_KEY_SECRET": COINBASE_API_KEY_SECRET,
        "COINBASE_WALLET_SECRET": COINBASE_WALLET_SECRET,
    },
)
print(f"✅ Credentials saved to {os.path.abspath(env_path)}")
print(f"   Added: {result.get('added', [])}")
print(f"   Updated: {result.get('updated', [])}")
print("\n   Never commit .env to git.")
```

## 6단계 — Delegated Signing 활성화

> ✋ **수동 작업 — Tutorial 00을 실행하기 전에 수행하세요. ProcessPayment가 성공하려면 Delegated Signing이 필요합니다.**

에이전트가 최종 사용자를 대신하여 transaction에 sign하려면 CDP project에서
delegated signing을 활성화해야 합니다. 한 번만 설정하는 project-level 항목입니다.

1. [CDP Portal](https://portal.cdp.coinbase.com/)을 엽니다.
2. project → **Wallets** → **Embedded Wallet**로 이동합니다.
3. **Policies** tab으로 이동합니다.
4. **Delegated Signing**을 활성화합니다.

![CDP Portal — Delegated Signing 활성화](../images/coinbase-setup/coinbase-setup-04.png)

활성화하면 이 project에서 생성된 모든 embedded wallet이 delegated signing을 지원합니다. wallet별 작업은 필요하지 않습니다.
